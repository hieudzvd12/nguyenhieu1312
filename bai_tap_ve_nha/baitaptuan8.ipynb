Cho một mảng số nguyên không âm. Gọi mảng đó là "mảng đẹp" nếu 
mỗi phần tử x trong mảng xuất hiện đúng x lần. 
Kiểm tra xem mảng có phải là mảng đẹp hay không. Nếu không hãy xóa 
đi y số (y là nhỏ nhất có thể) để mảng còn lại thành mảng đẹp. 

```python
def mang_dep(arr):
    xet = []
    for i in set(arr):
        dem = 0
        for j in arr:
            if i == j:
                dem = dem + 1
        if dem != i:
            xet.append(i)
    if xet == []:
        return 'la mang dep'
    else:
        so_can_xoa = 0
        for k in xet:
            dem_2 = 0
            for l in arr:
                if k == l:
                    dem_2 = dem_2 + 1
            if k > dem_2:
                so_can_xoa = so_can_xoa + dem_2
            elif k < dem_2:
                so_can_xoa = so_can_xoa + (dem_2 - k)
        return so_can_xoa
arr = list(map(int,input().split()))
kq = mang_dep(arr)
print(kq)
    
    
```

W7A17 
Tìm phần tử xuất hiện nhiều hơn n/2 lần trong mảng (nếu có) với độ 
phức tạp thời gian là O(n), độ phức tạp bộ nhớ là O(1), - Boyer–Moore. 
Nếu không có phần tử nào thỏa mãn thì không in gì 

```python
def tim_ptu(arr):
    n = len(arr)
    arr_2 = []
    for i in set(arr):
        dem = 0
        for j in arr:
            if i == j:
                dem = dem + 1
        if dem > n/2:
            arr_2.append(i)
    return arr_2
arr = list(map(int,input().split()))
kq = tim_ptu(arr)
print(kq)
```

Sắp xếp có điều kiện: Số chẵn đứng trước, số lẻ đứng sau, chẵn tăng 
dần, lẻ giảm dần; ;  

```python
day = list(map(int,input().split()))
day_chan = []
day_le = []
for i in day:
    if i % 2 == 0:
        day_chan.append(i)
    else:
        day_le.append(i)
for k in range(0,len(day_chan)):
    for j in range(k+1,len(day_chan)):
        if day_chan[k] > day_chan[j]:
            day_chan[k],day_chan[j] = day_chan[j],day_chan[k]
for l in range(0,len(day_le)):
    for m in range(l+1,len(day_le)):
        if day_le[l] < day_le[m]:
            day_le[l],day_le[m] = day_le[m],day_le[l]
daycuoi = day_chan + day_le
print(daycuoi)
```

```python
def tim_so_Ngay(N,T):
    thang_31 = [1,3,5,7,8,10,12]
    thang_30 = [4,6,9,11]
    tong_31 = 0
    tong_30 = 0
    if T == 2:
        return(31 + N)
    elif T == 1:
        return N
    else:
        for i in thang_31:
            if i < T:
                tong_31 = tong_31 + 31
        for j in thang_30:
            if j < T:
                tong_30 = tong_30 + 30
        return (tong_31 + tong_30 + N + 29)
N,T = map(int,input().split())
kq = tim_so_Ngay(N,T)
print(kq)


```

W7A25
Tìm số có đúng k chữ số lớn nhất có thể
trong một số có n chữ số (coi đầu vào là
một số rất lớn - str). Không được hoán
đổi vị trí các chữ số


```python
s = input()
k = int(input())
ans = ""
start = 0
for i in range(k):
    end = len(s) - (k - i) + 1
    mx = start
    for j in range(start, end):
        if s[j] > s[mx]:
            mx = j
    ans += s[mx]
    start = mx + 1
print(ans)

```

W7A27
Cho một dãy ngoặc ()(){}{}[][]..., kiểm tra
xem dã cho có hợp lệ hay không

```python
s = input()
stack = []
ok = True
for ch in s:
    if ch == "(" or ch == "[" or ch == "{":
        stack.append(ch)
    else:
        if len(stack) == 0:
            ok = False
            break
        top = stack.pop()
        if ch == ")" and top != "(":
            ok = False
        if ch == "]" and top != "[":
            ok = False
        if ch == "}" and top != "{":
            ok = False
if len(stack) != 0:
    ok = False
if ok:
    print("YES")
else:
    print("NO")
```

̀i toán Tìm dãy các hợp số dài nhất liên
tiếp không chứa số nguyên nào <= N (N
là số nguyên dương cho trước). Nếu có
nhiều dãy cùng độ dài thì lấy dãy nhỏ
nhất (gặp đầu tiên từ trái qua)

```python
def hop_so(n):
    if n < 4:
        return False
    dem = 0
    for i in range(1, n + 1):
        if n % i == 0:
            dem = dem + 1
    return dem > 2
N = int(input())
best_start = 0
best_len = 0
cur_start = 0
cur_len = 0
for i in range(1, N + 1):
    if hop_so(i):
        if cur_len == 0:
            cur_start = i

        cur_len = cur_len + 1
    else:
        if cur_len > best_len:
            best_len = cur_len
            best_start = cur_start
        cur_len = 0
if cur_len > best_len:
    best_len = cur_len
    best_start = cur_start
for i in range(best_start, best_start + best_len):
    print(i, end=" ")
```
